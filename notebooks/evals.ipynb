# Evals: does the tuned pipeline generalise?
The Jev hybrid's thresholds were tuned on the 60 Globex questions, so its score there is optimistic. This notebook runs the pipeline on **26 fresh questions** (`data/evals/heldout_set.csv`). They were written and committed to GitHub before any run, so nothing was tuned to them.

What it measures:
- **Unsafe auto-approvals:** must be 0.
- **Answerable questions auto-approved**, and the **precision of High** answers.
- **Traps escalated** and **restricted questions held back**.
- **Routing** and **citation coverage**.
- **Faithfulness:** an LLM judge (Claude Opus, a different model from the Sonnet drafter) checks every statement against the cited sections.
